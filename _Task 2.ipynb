```python
import pandas as pd
df = pd.read_csv(r"D:\Ugin Venasa\Titanic-Dataset.csv")
df.head()
```

```python
df.describe(include='all')
df.info()
df.isnull().sum()
```

```python
df.hist(bins=20, figsize=(12, 8), color='skyblue')
plt.suptitle('Distribution of Numerical Features')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='Pclass', y='Age')
plt.title('Boxplot of Age by Passenger Class')
plt.show()
sns.boxplot(data=df, y='Fare')
plt.title('Boxplot of Fare')
plt.show()
```

```python
df_encoded = df.copy()
df_encoded['Sex'] = df_encoded['Sex'].map({'male': 0, 'female': 1})
df_encoded['Embarked'] = df_encoded['Embarked'].map({'S': 0, 'C': 1, 'Q': 2})
non_numeric_cols = ['Name', 'Ticket', 'Cabin']  
df_encoded = df_encoded.drop(columns=non_numeric_cols)
plt.figure(figsize=(10, 6))
sns.heatmap(df_encoded.corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()
```

```python
sns.pairplot(df[['Survived', 'Age', 'Fare', 'Pclass', 'SibSp', 'Parch']], hue='Survived')
plt.suptitle('Pairplot of Selected Features')
plt.show()
```
